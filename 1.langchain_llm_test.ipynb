```python
from langchain_upstage import ChatUpstage

llm = ChatUpstage(model="solar-mini4")
```

```python
ai_message = llm.invoke("인프런에 어떤 강의가 있나요?")
print(ai_message.content)
```

```python
from langchain_upstage import ChatUpstage

llm = ChatUpstage(
    model="solar-mini4",
    timeout=20,
    max_retries=0
)

response = llm.invoke("안녕")
print(response.content)
```
